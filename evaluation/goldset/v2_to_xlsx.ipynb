# v2 JSON → 검수용 xlsx 변환

`batch_label_v2.py`가 생성한 `*_v2.json` 파일을 읽어
**`human_check` 열이 있는 xlsx**로 변환한다.

- 표기법: `O`=맞음 / `X`=아님 / `M`=나눠야 / `S`=합쳐야
- 검수 후 `label_goldset.ipynb`의 확정 셀 실행

```python
import json
from pathlib import Path

import openpyxl
import pandas as pd
from openpyxl.styles import Alignment

GS = Path("../data/goldset/labeling")
DATES = ["2026-02-10", "2026-02-11", "2026-02-12", "2026-02-13"]


def _write_xlsx(path: Path, rows: list[dict], col_widths: dict) -> None:
    pd.DataFrame(rows).to_excel(path, index=False)
    wb = openpyxl.load_workbook(path)
    ws = wb.active
    for col, w in col_widths.items():
        ws.column_dimensions[col].width = w
    for row in ws.iter_rows(min_row=2):
        for cell in row:
            cell.alignment = Alignment(wrap_text=True, vertical="top")
    wb.save(path)


print(f"대상 날짜: {DATES}")
```

```python
for date in DATES:
    print(f"\n{'='*50}")
    print(f"날짜: {date}")

    # ── 개념 ──────────────────────────────────────────
    co_path = GS / f"{date}_concept_goldset_v2.json"
    if co_path.exists():
        co = json.loads(co_path.read_text(encoding="utf-8"))
        rows = [
            {"timestamp": c["timestamp"], "concept_name": c["concept_name"],
             "span": f"{c['start']}~{c['end']}", "key_sentence": c["key_sentence"],
             "text": c["text"], "human_check": ""}
            for c in co["concepts"]
        ]
        out = GS / f"{date}_concept_goldset_v2.xlsx"
        _write_xlsx(out, rows, {"A": 12, "B": 28, "C": 10, "D": 50, "E": 70, "F": 12})
        print(f"  개념  {len(rows)}개 → {out.name}")
    else:
        print(f"  개념  [없음] {co_path.name}")

    # ── 예시 ──────────────────────────────────────────
    ex_path = GS / f"{date}_example_goldset_v2.json"
    if ex_path.exists():
        ex = json.loads(ex_path.read_text(encoding="utf-8"))
        rows = [
            {"timestamp": e["timestamp"], "example_name": e["example_name"],
             "span": f"{e['start']}~{e['end']}", "key_sentence": e["key_sentence"],
             "text": e["text"], "human_check": ""}
            for e in ex["examples"]
        ]
        out = GS / f"{date}_example_goldset_v2.xlsx"
        _write_xlsx(out, rows, {"A": 12, "B": 30, "C": 12, "D": 45, "E": 70, "F": 12})
        print(f"  예시  {len(rows)}개 → {out.name}")
    else:
        print(f"  예시  [없음] {ex_path.name}")

    # ── 실습 ──────────────────────────────────────────
    pr_path = GS / f"{date}_practice_goldset_v2.json"
    if pr_path.exists():
        pr = json.loads(pr_path.read_text(encoding="utf-8"))
        rows = [
            {"timestamp": p["timestamp"], "practice_name": p["practice_name"],
             "span": f"{p['start']}~{p['end']}", "key_sentence": p["key_sentence"],
             "text": p["text"], "human_check": ""}
            for p in pr["practices"]
        ]
        out = GS / f"{date}_practice_goldset_v2.xlsx"
        _write_xlsx(out, rows, {"A": 12, "B": 30, "C": 12, "D": 45, "E": 70, "F": 12})
        print(f"  실습  {len(rows)}개 → {out.name}")
    else:
        print(f"  실습  [없음] {pr_path.name}")

print("\n완료.")
```
